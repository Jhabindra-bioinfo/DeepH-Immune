# 02 — Train DeepH-Immune with five-fold cross-validation

Run notebook 01 first to generate training ESM-2 features.


```
from pathlib import Path
import os, sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent

assert (ROOT / "src").exists(), "Run from the cloned repository or notebooks/ directory."
print("Repository root:", ROOT)
```

```
import subprocess, sys

cmd = [
    sys.executable,
    str(ROOT / "src" / "train_deeph_immune.py"),
    "--csv", str(ROOT / "data" / "training" / "Training_MHCII_V1_to_V2_12_15_clean.csv"),
    "--esm-dir", str(ROOT / "data" / "training" / "esm2_features"),
    "--outdir", str(ROOT / "results" / "cross_validation"),
    "--split-mode", "stratified_length",
    "--folds", "5",
    "--epochs", "80",
    "--batch-size", "64",
    "--seed", "42",
]

print(" ".join(cmd))
subprocess.run(cmd, cwd=ROOT, check=True)
```
