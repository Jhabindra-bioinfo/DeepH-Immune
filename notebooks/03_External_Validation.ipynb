# 03 — Reproduce external validation

The external ESM-2 features are distributed with the repository so this analysis can be run directly without recomputing ESM-2 embeddings.


```
from pathlib import Path
import os, sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent

assert (ROOT / "src").exists(), "Run from the cloned repository or notebooks/ directory."
print("Repository root:", ROOT)
```

```
import subprocess, sys

cmd = [
    sys.executable,
    str(ROOT / "external_validation" / "run_external_validation.py"),
]

print(" ".join(cmd))
subprocess.run(cmd, cwd=ROOT, check=True)
```
