# 01 — Generate training ESM-2 features

The repository intentionally does **not** distribute precomputed training ESM-2 features.

This notebook generates them from the released model-ready training CSV using frozen `facebook/esm2_t33_650M_UR50D`.


```
from pathlib import Path
import os, sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent

assert (ROOT / "src").exists(), "Run from the cloned repository or notebooks/ directory."
print("Repository root:", ROOT)
```

```
import subprocess, sys

cmd = [
    sys.executable,
    str(ROOT / "src" / "extract_esm2.py"),
    "--csv", str(ROOT / "data" / "training" / "Training_MHCII_V1_to_V2_12_15_clean.csv"),
    "--outdir", str(ROOT / "data" / "training" / "esm2_features"),
    "--model", "facebook/esm2_t33_650M_UR50D",
    "--peptide-batch-size", "128",
    "--mhc-batch-size", "16",
]

print(" ".join(cmd))
subprocess.run(cmd, cwd=ROOT, check=True)
```
